#### Query a Foundation Model 

Let's use the difference between "SYSTEM" and "USER" Chat Message Roles to send different prompts to the AI agent based on the Chat Message role.

`Tell the AI Agent to act like a high school computer science teacher when answering queries using the prompt ai_agent_context.`
> - Ask the AI agent when to use For loops using the user_prompt.

```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service.serving import ChatMessage, ChatMessageRole

# "databricks-meta-llama-3-3-70b-instruct" is the name of a Large Language Model (LLM) serving endpoint in Databricks. Your Python code sends a prompt to this endpoint and receives an AI-generated response.



user_prompt = "What is a For loop and When should I use it?"
ai_agent_context = "You are a high school computer science teacher"
response = w.serving_endpoints.query(
    name="databricks-meta-llama-3-3-70b-instruct",

# This code sends a system prompt to a Databricks-hosted Llama 3.3 70B model, defining its role as a high school computer science teacher. The agent context guides the model’s response to the user’s question about Python `for` loops.

    messages=[
        ChatMessage( 
          	# Set system prompt
            role=ChatMessageRole.SYSTEM, content=ai_agent_context
        ),

# This code adds the user's question as a `USER` message and sets `max_tokens=250` to limit the model's response length. The model generates an answer using both the user's query and the previously defined system prompt.

        ChatMessage( 
          	# Set user query
            role=ChatMessageRole.USER, content=user_prompt
        ),
    ],
    max_tokens=250)
print(f"RESPONSE:\n{response.choices[0].message.content}")
```

#### Query chatbot using Databricks Servings Endpoint API


In the previous exercise, you learned how to use the Databricks SDK to query an AI Foundation model. 
- The goal of this exercise is for you to experiment with `different user and system prompts` when querying AI models using the Databricks SDK

> Edit the System prompt and see how the AI model responds differently.
> Edit the User prompt and see how the AI model responds differently.

```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service.serving import ChatMessage, ChatMessageRole
w = WorkspaceClient()
response = w.serving_endpoints.query(
    name="databricks-meta-llama-3-3-70b-instruct",
    messages=[
        ChatMessage( 
          	# Edit the System prompt
            role=ChatMessageRole.SYSTEM, content="You are good python programmer for developer"

# This code updates the system prompt to define the model as an expert Python programmer for developers, guiding it to provide programming-focused responses when processing subsequent user queries.

        ),
        ChatMessage( 
          	# Edit the User prompt
            role=ChatMessageRole.USER, content="How Python loop works in SDK?" 
        ),
    ],
    max_tokens=128)

# This code sends the user’s question, “How Python loop works in SDK?”, along with the system prompt to the Databricks-hosted Llama model. The `max_tokens=128` parameter limits the generated response to 128 tokens.

print(f"RESPONSE:\n{response.choices[0].message.content}")
```

#### Text summarization

- Want to turn a poem into a quick summary? 
Let’s make it happen! You have the full text of a poem stored as a string variable called raven_text, which you will hand off to a Chat AI model and ask it to summarize the poem in a set number of words.

`Ask the AI agent to summarize the poem "The Raven" by Edgar Allan Poe.`
> Define the model name as Meta Llama 3, already saved as ai_model. Create the message you want to send to the AI model, including the raven_text variable.
> Use the chat message role used to query the AI model.

```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service.serving import ChatMessage, ChatMessageRole

w = WorkspaceClient()

ai_model = "databricks-meta-llama-3-3-70b-instruct"

raven_text = "The Raven by Edgar Allan Poe"

response = w.serving_endpoints.query(
    name=ai_model,
    messages=[
        ChatMessage(
            role=ChatMessageRole.USER,
            content=f"Summarize this poem by Edgar Allan Poe: {raven_text} in 3 bullet points"
        )
    ],
    max_tokens=200
)

# This code queries the Databricks-hosted Llama 3.3 70B model to summarize The Raven by Edgar Allan Poe in three bullet points, using an f-string to insert the poem title into the prompt and `max_tokens=200` to limit the response length.

print(f"RESPONSE:\n{response.choices[0].message.content}")
```

#### Proofreading text

- Have text that you want to proofread? 
Use the Databricks SDK to ask an AI model to proofread your text for you. 
`It will send you back the proofread version of your text and a list of the specific changes it made.`

> - Query the class in the workspace client used to query AI models.
> - Fill in the role used to provide instructions to the AI model on how to answer future queries.
> - Fill in the role used when sending a query to the AI model.
> - Print the message details from the AI model response.

```python
w = WorkspaceClient()

text_to_proofread = "I am gong to the stor to buy some bred. I migt also buy some mlk. Do u want to jon me?"

# Query the AI model via the workspace client
response = w.serving_endpoints.query(
    name="databricks-meta-llama-3-3-70b-instruct",
    messages=[
        ChatMessage(
            # Specify the role for configuring model behavior
            role=ChatMessageRole.SYSTEM,
            content="You are a helpful writing assistant."
        ),
# This code defines a system message that configures the Databricks-hosted Llama model to act as a helpful writing assistant, guiding its behavior and response style for subsequent user prompts.

        ChatMessage(
            # Specify the role for sending the query
            role=ChatMessageRole.USER,
            content=f"Proofread the following text: {text_to_proofread}"
        ),
    ],
    max_tokens=200
)

# This code sends the text stored in `text_to_proofread` to the model with a USER message asking it to proofread the text. The `max_tokens=200` parameter limits the length of the generated response.

# Print the content of the model's response
print(f"RESPONSE:\n{response.choices[0].message.content}")
```
